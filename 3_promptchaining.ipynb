```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
gemini_api_key = os.getenv("GEMINI_API_KEY")
gemini_model = os.getenv("GEMINI_MODEL", "gemini-2.5-flash")

model = ChatOpenAI(
    model=gemini_model,
    api_key=gemini_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai",
)
```

```python
#start->genoutline->gencontent->end
```

```python
#define state schema
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
    
```

```python
def gen_outline(state: BlogState) -> BlogState:
    title=state['title']
    
    prompt = f"Generate a detailed outline for a blog post with the title: {title}"
    outline = model.invoke(prompt).content
    
    state['outline'] = outline
    return state
```

```python
def gen_content(state: BlogState) -> BlogState:
    title=state['title']
    outline=state['outline']
    
    prompt = f"Generate a detailed blog post with the title: {title} using the outline: {outline}"
    content = model.invoke(prompt).content
    
    state['content'] = content
    return state
```

```python
#define graph
graph=StateGraph(BlogState)
#nodes
graph.add_node("gen_outline",gen_outline)
graph.add_node("gen_content",gen_content)

#edges
graph.add_edge(START,"gen_outline")
graph.add_edge("gen_outline","gen_content")
graph.add_edge("gen_content",END)

#compile graph
workflow=graph.compile()
```

```python
#execute
intial_state=BlogState(title="The Future of AI")
final_state=workflow.invoke(intial_state)
print(final_state)
```
